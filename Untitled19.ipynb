```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

1. Download the Heart Disease dataset from Kaggle.
2. Load the dataset using Pandas.

```python
df = pd.read_csv(r"D:\Downloaddd\heart.csv")
print(df)
```

3. Explore the dataset: display the first five rows, number of rows and columns, input features and target
variable, and check for missing values.


```python
print("--- First 5 Rows ---")
print(df.head())

print("\n--- Shape (Rows, Columns) ---")
print(df.shape)

target_col = "target"
features = [col for col in df.columns if col != target_col]
print("\n--- Input Features ---")
print(features)
print("\n--- Target Variable ---")
print(target_col)
print("\n--- Missing Values ---")
print(df.isnull().sum())
```

4. Perform the required preprocessing.

```python
df = df.dropna()
df = pd.get_dummies(df, drop_first=True)
```

5. Select input features and target variable.

```python
X = df.drop(columns=[target_col])
y = df[target_col]
```

6. Split the dataset into training and testing sets using test_size = 0.30.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42
)
```

7. Apply feature scaling using StandardScaler.

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

8. Implement KNN classification using KNeighborsClassifier.
9. Train the model using fit() and generate predictions using predict().
10. Calculate and display model accuracy.

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_scaled, y_train)
y_pred = knn.predict(X_test_scaled)
acc = accuracy_score(y_test, y_pred)
print(f"\nAccuracy with K=5: {acc:.4f} ({acc * 100:.2f}%)")
```

11. Implement and compare KNN using K = 3, 5, 7, and 9.
12. Compare the accuracies and identify the best-performing K value.

```python
k_values = [3, 5, 7, 9]
accuracies = {}

print("\n--- Comparing K Values ---")
for k in k_values:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train_scaled, y_train)
    preds = model.predict(X_test_scaled)
    score = accuracy_score(y_test, preds)
    accuracies[k] = score
    print(f"K = {k} -> Accuracy: {score:.4f} ({score * 100:.2f}%)")

best_k = max(accuracies, key=accuracies.get)
best_acc = accuracies[best_k]
print(f"\nBest performing K value is K = {best_k} with accuracy {best_acc * 100:.2f}%")
```

Conceptual Questions
1. What is KNN?
   KNN (K-Nearest Neighbors) is a simple supervised machine learning algorithm used for classification and regression. It predicts the group of a new     data point by looking at the existing data points closest to it.
   


2. How does KNN classify a new data point?
   When a new data point is given, KNN measures its distance from all the training data points. It picks the closest K neighbors and checks their         labels. The label that appears most often among those neighbors (majority voting) is given to the new point.

3. What is the role of K in KNN?
   K decides how many nearest neighbors the model will look at before making a decision. For example, if K = 5, the model checks the 5 closest points     to decide the final answer.

4. Why is feature scaling necessary for KNN?
   KNN relies on distance to find the nearest neighbors. If one feature has large numbers (like cholesterol around 250) and another has small numbers     (like age around 50), the large numbers will dominate the distance calculation. Feature scaling puts all features on the same scale so every           feature has a fair impact.

5. What is Euclidean distance?
   Euclidean distance is simply the straight-line distance between two points. In KNN, it is the most common way to measure how close or far two data     points are from each other.

6. How can changing K affect model predictions?
   If K is too small (like K = 1), the model becomes too sensitive to noise and can overfit the data. If K is too large, the model looks at too many      points and may miss important local patterns (underfitting). Testing different values like 3, 5, 7, and 9 helps find the best balance.
